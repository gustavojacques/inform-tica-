```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("raphaelmanayon/temperature-and-ice-cream-sales")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os

# The dataset is located at 'path' and the file is 'Ice Cream.csv'
data_filepath = os.path.join(path, 'Ice Cream Sales - temperatures.csv')
df = pd.read_csv(data_filepath)
```

```python
# Display the first 5 rows of the DataFrame to confirm loading
display(df.head())
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Define the independent variable (X) and dependent variable (y)
# Reshape X to be a 2D array as required by scikit-learn
X = df['Temperature'].values.reshape(-1, 1)
y = df['Ice Cream Profits'].values
```

```python
# Create and train the Linear Regression model
model = LinearRegression()
model.fit(X, y)

# Display the model's coefficients
print(f"Coeficiente (inclinação): {model.coef_[0]:.2f}")
print(f"Intercepto: {model.intercept_:.2f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a scatter plot of the actual data points
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Temperature', y='Ice Cream Profits', data=df, label='Dados Reais')

# Plot the regression line
# Generate predictions for the range of X values
plt.plot(X, model.predict(X), color='red', label='Linha de Regressão Linear')

plt.title('Vendas de Sorvete vs. Temperatura com Regressão Linear')
plt.xlabel('Temperatura (°F)')
plt.ylabel('Lucro com Sorvete ($)')
plt.legend()
plt.grid(True)
plt.show()
```
